## problem 1 Word frequency Counter
take  a sentence as input

Display how many times each words 

Eg1: python is fun python is easy

op : python: 2 is, :2 ,easy: 2 ,fun:1

```python
s = input()
words= s.split()
print(words)
```

```python
freq = {}
for word in words:
    if word not in freq:
        freq[word]= 1
    else:
        freq[word]+=1
    print(word)
```

## problem 3 Chracter counter
take a string input from the user

Display

Number of vowel 

number of constants

number of digits

number of space

ip : python is fun 

op : vowls 3
constant 8 or digit 1 or spaces 2

```python
passward = input("enter the passward: ")
vowel = 0
constant = 0
digits = 0
space = 0
for c in passward:
    if c in "aeiou":
        vowel+=1
    elif c>="a" and  c<="z":
        constant+=1
    elif c>='0'  and c<='9':
        digits+=1
    elif c ==" ":
        space+=1
print(passward)
print("vowels",vowel)
print("constant",constant)
print("digits",digits)
print("space",space)
```

## problem 5 passward genrater


Problem Statement 

A password is valid if it contains 
minimum 8 characters 

at least one uppercase letter 

at least one lowercase letter

at least one digit 

Display Strong 

Password or Weak Password

```python
passward = "Adity2333"
upper = False
lower = False
digits = False
n = len(passward)
for c in passward:
    if c.islower():
        lower=True
    elif c.isupper():
        upper=True
    elif c.isdigit():
        digits=True
if n>=8 and upper==True and lower==True and digits==True:
    print("Strong passward")
else:
    print("weak passward")
```
