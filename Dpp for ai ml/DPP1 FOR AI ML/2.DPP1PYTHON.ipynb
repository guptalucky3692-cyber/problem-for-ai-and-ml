```python
class bank:
    def __init__(self,balance):
        self.balance = balance
    def withdrow(self,amount):
        self.amount = amount
        if self.amount<= self.balance:
            self.balance-=self.amount
            print(f"your cash  will be sucess fully in this {amount}")
account = bank(1000)
account.withdrow(300)
print(account.balance)
        
```

## 3.lambda and Map filter DPP solution

```python
num =[10,12,13]
result = list(map(lambda x:x // 5,num))
print(result)
```

```python
num2 = [3,6,9,12,36]
result =list(map(lambda x:x+1,filter(lambda x: x%6==0,num2)))
print(result)
```

```python
marks = [22,21,34,56,43]
passed = filter(lambda x:x>=40,marks)
bonus = list(map(lambda x:x+5,passed))
print(bonus)
```

```python
numbers =[2,3,4]
result = list(map(lambda x:x**2,filter(lambda x:x>2,numbers)))
print(result)
```

```python
from functools import reduce
numbers =[1,2,3,4,5]
result=list(map(lambda x:x*10,filter(lambda x:x%2!=0,numbers)))
print(result)
```
